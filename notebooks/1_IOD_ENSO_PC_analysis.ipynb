```python
def eof_indian(x):
    
    sl = x.copy()
    
    if 'time_counter' in sl.coords:
        sl = sl.rename ( {'time_counter':'time'} )
    else:
        pass
        
    sl = sl.sel(lat=slice(-21,21))
    
    #sl_indian_clim = sl_indian.groupby('time.month').mean(dim='time')
    #sl_indian = sl_indian.groupby('time.month') - sl_indian_clim
    
    #sl_indian = detrend(sl_indian.fillna(0), 'time', detrend_type="linear")

    y = sl.sel(lat=slice(-21,21)).coords['lat'].values
    coslat = np.cos(np.deg2rad(y))
    wgts = np.sqrt(coslat)[...,np.newaxis]
    
    solver = Eof(sl,weights=wgts)

    eof = solver.eofs(neofs=3)
    pc = solver.pcs(npcs=3)
    
    pc1 = pc.sel(mode=0)
    
    pc1 = pc1/pc1.std()
    
    pc2 = pc.sel(mode=1)

    pc2 = pc2/pc2.std()
    
    pc3 = pc.sel(mode=2)
    
    pc3 = pc3/pc3.std()
    
    eof1 = eof.sel(mode=0)
    eof2 = eof.sel(mode=1)
    eof3 = eof.sel(mode=2)
    
    var_expl = solver.varianceFraction(neigs=3)
    
    return pc1, pc2, pc3, eof1, eof2, eof3, var_expl


def lag_linregress_3D(x, y, time='time',lagx=0, lagy=0):
    """
    Input: Two xr.Datarrays of any dimensions with the first dim being time. 
    Thus the input data could be a 1D time series, or for example, have three 
    dimensions (time,lat,lon). 
    Datasets can be provided in any order, but note that the regression slope 
    and intercept will be calculated for y with respect to x.
    Output: Covariance, correlation, regression slope and intercept, p-value, 
    and standard error on regression between the two datasets along their 
    aligned time dimension.  
    Lag values can be assigned to either of the data, with lagx shifting x, and
    lagy shifting y, with the specified lag amount. 
    """ 
    #1. Ensure that the data are properly alinged to each other. 
    x,y = xr.align(x,y)

    #2. Add lag information if any, and shift the data accordingly
    if lagx!=0:

        # If x lags y by 1, x must be shifted 1 step backwards. 
        # But as the 'zero-th' value is nonexistant, xr assigns it as invalid 
        # (nan). Hence it needs to be dropped
        x   = x.shift(time = -lagx).dropna(dim=time)

        # Next important step is to re-align the two datasets so that y adjusts
        # to the changed coordinates of x
        x,y = xr.align(x,y)

    if lagy!=0:
        y   = y.shift(time = -lagy).dropna(dim=time)
        x,y = xr.align(x,y)

    #3. Compute data length, mean and standard deviation along time axis: 
    n = y.notnull().sum(dim=time)
    xmean = x.mean(axis=0)
    ymean = y.mean(axis=0)
    xstd  = x.std(axis=0)
    ystd  = y.std(axis=0)

    #4. Compute covariance along time axis
    cov   =  ((x - xmean)*(y - ymean)).sum(axis=0)/(n)

    #5. Compute correlation along time axis
    cor   = cov/(xstd*ystd)

    #6. Compute regression slope and intercept:
    slope     = cov/(xstd**2)
    intercept = ymean - xmean*slope  

    #7. Compute P-value and standard error
    #Compute t-statistics
    tstats = cor*np.sqrt(n-2)/np.sqrt(1-cor**2)
    stderr = slope/tstats

    from scipy.stats import t
    pval   = t.sf(tstats, n-2)*2
    pval   = xr.DataArray(pval, dims=cor.dims, coords=cor.coords)
    
    return cov,cor,slope,intercept,pval,stderr


def poly(x,y,color='k',ax=None):
    poly_corners = np.zeros((len(y), 2), np.float64)
    poly_corners[:,0] = x
    poly_corners[:,1] = y

    poly = mpatches.Polygon(poly_corners, closed=True, ec=color, fill=False,lw=2, transform=ccrs.PlateCarree())

    return ax.add_patch(poly)

def anom(ds):
    
    ds_clim = ds.groupby('time.month').mean(dim='time')
    ds_anom = ds.groupby('time.month') - ds_clim
    
    return ds_anom

def NINO34_idx(sst,time='time'):
    
    if sst.lat.values[0] < 0:
        sst = sst.sel(lat=slice(-6,6),lon=slice(360-170,360-120))
    else:
        sst = sst.sel(lat=slice(6,-6),lon=slice(360-170,360-120))
        
    nino34 = sst.mean(dim=('lat','lon'))
    
    return anom(nino34)

def NINO34(sst,model=None,time='time'):
    
    if sst.lat.values[0] < 0:
        sst = sst.sel(lat=slice(-6,6),lon=slice(360-170,360-120))
    else:
        sst = sst.sel(lat=slice(6,-6),lon=slice(360-170,360-120))

    return sst.mean(dim=('lat','lon'))


def EIO(sst,model=None,time='time'):
    
    if sst.lat.values[0] < 0:
        sst_EIO = sst.sel(lat=slice(-11,0),lon=slice(90,110))
    else:
        sst_EIO = sst.sel(lat=slice(0,-11),lon=slice(90,110))

    return sst_EIO.mean(dim=('lat','lon'))


def WIO(sst,model=None,time='time'):
    
    if sst.lat.values[0] < 0:
        sst_WIO = sst.sel(lat=slice(-11,11),lon=slice(50,70))
    else:
        sst_WIO = sst.sel(lat=slice(11,-11),lon=slice(50,70))

    return sst_WIO.mean(dim=('lat','lon'))

def DMI(sst,model=None,time='time'):
    
    sst_DMI = WIO(sst) - EIO(sst)
    
    return sst_DMI

def DMI_idx(ds):
    return anom(WIO(ds)) - anom(EIO(ds))
```

## the ENSO → Indian Ocean bridge in the model and in observations
Quantifies what the regression maps above (Fig. S3) show qualitatively: the SON Indian Ocean response (WIO, EIO, DMI,
equatorial zonal wind stress, Maritime-Continent rainfall) per standard deviation of the SON Niño3.4 index, in
observations (1950–2019, detrended) and in CESM1.2 (last 250 years, all 250-year windows of the 2 Ma, and the two
perihelion composites of Fig. 3). Box means from the calendar-adjusted fields (notebook 3).

```python
import pickle
from scipy import stats
# ENSO → Indian Ocean "atmospheric bridge": how much of the Indian Ocean response to Niño3.4 does CESM1.2
# reproduce? Complements the regression maps above (Fig. S3) with basin-mean numbers. SON anomalies of WIO,
# EIO and DMI (and of the equatorial Indian Ocean zonal wind stress and Maritime-Continent rainfall in the
# model) are regressed on the standardized SON Niño3.4 index: in observations (1950–2019, detrended) and in
# the model for the last 250 years, for all 250-year windows of the 2 Ma, and for the two perihelion
# composites of Fig. 3. Box means come from the calendar-adjusted fields (notebook 3).
boxes = xr.open_dataset('variables/derived_data/boxes_monthly_adjusted.nc')
vals = pickle.load(open('variables/vals_new.pkl', 'rb'))
params_ = xr.open_dataset('variables/orbital_params/pCESM.2Ma_orbPara.nc')
W = 250; NY = boxes.sizes['year']; nw = NY // W
centre_ka = -(NY - (np.arange(nw) * W + W / 2)) * 5 / 1000
lop_w = params_.interp(year=centre_ka)['longitude_of_perihelion'].values % 360
sel_n34 = (lop_w >= vals.loc['Niño3.4', 'Range_high'][0]) & (lop_w <= vals.loc['Niño3.4', 'Range_high'][1])
sel_dmi = (lop_w >= vals.loc['DMI', 'Range_high'][0]) & (lop_w <= vals.loc['DMI', 'Range_high'][1])

def son(x): return x[:, 8:11].mean(axis=1)
def bridge_stats(series):
    """series: dict of (year, month) anomaly arrays with keys n34, wio, eio, [taux_io, prect_mc]."""
    z = son(series['n34']); z = (z - z.mean()) / z.std()
    out = {'σ Niño3.4 SON [°C]': son(series['n34']).std()}
    for k, lab, f in [('wio', 'WIO [°C/σ]', 1), ('eio', 'EIO [°C/σ]', 1), ('dmi', 'DMI [°C/σ]', 1),
                      ('taux_io', 'eq. IO zonal stress [10⁻³ N/m²/σ]', -1e3), ('prect_mc', 'Maritime Cont. rain [mm/day/σ]', 8.64e7)]:
        if k == 'dmi': y = son(series['wio'] - series['eio'])
        elif k in series: y = son(series[k])
        else: out[lab] = np.nan; continue
        out[lab] = stats.linregress(z, y).slope * f
    out['r(DMI, Niño3.4) SON'] = np.corrcoef(z, son(series['wio'] - series['eio']))[0, 1]
    return out

def model_series(sl):
    def a(k): x = boxes[k].values[sl].astype(float); return x - x.mean(axis=0, keepdims=True)
    return {'n34': a('TS_nino34'), 'wio': a('TS_wio'), 'eio': a('TS_eio'), 'taux_io': a('TAUX_eq_io'), 'prect_mc': a('PRECT_mc')}

def obs_series(sst):
    desc = sst.lat.values[0] > sst.lat.values[-1]
    def box(la0, la1, lo0, lo1):
        return sst.sel(lat=slice(la1, la0) if desc else slice(la0, la1), lon=slice(lo0, lo1)).mean(('lat', 'lon'))
    out = {}
    for k, x in [('n34', box(-5, 5, 190, 240)), ('wio', box(-10, 10, 50, 70)), ('eio', box(-10, 0, 90, 110))]:
        an = (x.groupby('time.month') - x.groupby('time.month').mean('time')).values
        t = np.arange(an.size); out[k] = (an - np.polyval(np.polyfit(t, an, 1), t)).reshape(-1, 12)
    return out

rows = {}
for name, f in [('COBE-SST2 1950–2019', 'variables/reference_data/sst_cobe.nc'), ('ERSSTv5 1950–2019', 'variables/reference_data/ersstv5.nc')]:
    rows[name] = bridge_stats(obs_series(xr.open_dataset(f)['sst'].sel(time=slice('1950-01-01', '2019-12-31'))))
rows['CESM1.2 last 250 years'] = bridge_stats(model_series(slice(NY - W, NY)))
per_window = pd.DataFrame([bridge_stats(model_series(slice(w * W, (w + 1) * W))) for w in range(nw)])
rows['CESM1.2 mean of 2 Ma windows'] = per_window.mean().to_dict()
rows['CESM1.2 max Niño3.4 σ phase'] = per_window[sel_n34].mean().to_dict()
rows['CESM1.2 max DMI σ phase'] = per_window[sel_dmi].mean().to_dict()
bridge = pd.DataFrame(rows).T
per_window['lon_perihelion'] = lop_w
per_window.to_csv('variables/derived_data/bridge_regressions_250yr.csv', index_label='window')
pd.set_option('display.width', 200)
print(bridge.round(3).to_string())
```

```python
# Figure: spread of the bridge statistics across the 2 Ma windows against the observed values. Colour code =
# orbital configuration (grey: all windows; blue: max Niño3.4 variability; orange: max DMI variability), as in Fig. 3;
# boxes: median, interquartile range and 5–95 % range; observations 1950–2019: star (COBE-SST2), cross (ERSSTv5).
import matplotlib.axes
C_BOX = ['0.6', '#1f77b4', '#ff7f0e']; LAB_BOX = ['all\nwindows', 'ENSO-\nfavouring', 'IOD-\nfavouring', 'obs.']
cols = [('WIO [°C/σ]', 'WIO SST'), ('EIO [°C/σ]', 'EIO SST'), ('DMI [°C/σ]', 'DMI'), ('r(DMI, Niño3.4) SON', 'DMI–Niño3.4 correlation')]
pplt.rc['font.size'] = 13
fig, axs = pplt.subplots([[1, 2], [3, 4]], refwidth=3.1, refheight=2.7, share=False, wspace=5, hspace=5)
for ax, (c, title) in zip(axs, cols):
    data = [per_window[c].values, per_window[c].values[sel_n34], per_window[c].values[sel_dmi]]
    bp = matplotlib.axes.Axes.boxplot(ax, data, positions=[0, 1, 2], widths=0.55, whis=(5, 95), showfliers=False, patch_artist=True,
                                      medianprops={'color': 'k', 'linewidth': 2})
    for patch, col in zip(bp['boxes'], C_BOX): patch.set_facecolor(col); patch.set_alpha(0.75)
    for name, mk in [('COBE-SST2 1950–2019', '*'), ('ERSSTv5 1950–2019', 'P')]:
        ax.scatter(3, bridge.loc[name, c], c='k', s=220, marker=mk, edgecolors='w', linewidths=0.8, zorder=10, label=name.split(' ')[0])
    ax.axhline(0, linestyle='--', color='k', alpha=0.4, linewidth=1)
    unit = 'per σ of Niño3.4 [°C]' if 'σ' in c else 'correlation'
    ax.format(title=title, ylabel=unit, xlocator=[0, 1, 2, 3], xformatter=LAB_BOX, xlim=(-0.6, 3.6), xtickminor=False, ytickminor=False, xticklabelsize=10)
axs[3].legend(loc='ul', frameon=False, fontsize=9)
axs.format(abc='a.', abcloc='ul', titleloc='c', titlesize=12, xtickdir='out', ytickdir='out', grid=False, ticklen=4,
           suptitle='September–November Indian Ocean response to Niño3.4')
plt.savefig('fig/bridge_model_obs.png', dpi=300, format='png', bbox_inches='tight', transparent=True)
plt.show()
```
